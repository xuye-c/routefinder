```python
import os
import re
import numpy as np
import pandas as pd
```

**Go through \results folder**

```python
root_dir = "results/solutions"
rows = []
pattern = re.compile(r"(.+?)_(\d+)_sol\.npz")
for root, dirs, files in os.walk(root_dir):
    for file in files:
        if file.endswith(".npz"):
            match = pattern.match(file)
            if match:
                variant = match.group(1)
                node_size = int(match.group(2))
                file_path = os.path.join(root, file)
                data = np.load(file_path)
                model_name = os.path.basename(root)

                costs = data["costs"]        # shape (1000,)
                times = data["time"]         # shape (1000, 113) or (1000,)

                avg_cost = costs.mean()

                rows.append({
                    "variant": variant,
                    "model": model_name,
                    "nodes": node_size,
                    "avg_cost": avg_cost,
                    "time": times,
                })

df = pd.DataFrame(rows)
df = df.sort_values(by=["variant", "model"])
print(df)
```

```python
df.to_csv("summary50.csv", index=False)

print("Done. Saved to summary.csv")
```
