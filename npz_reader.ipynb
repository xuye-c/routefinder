```python
def read(file_path):
    data = np.load(file_path)
    print("columns:", data.files)
    return data
def see_content(data, column_name):
    x = data[column_name]
    print("shape:", x.shape)
    print("dtype:", x.dtype)
    #print("first 5 rows:", x[:5])
    print("last 5 rows:", x[-5:])
```
